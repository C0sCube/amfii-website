```python
import os
import json
import fnmatch
import pandas as pd

def load_and_concat(path: str, pattern: str = "*.json", json_key: str = None) -> pd.DataFrame:
    """
    Walk through folder/subfolders, read matching .csv/.xlsx/.json files,
    and concatenate them into a single DataFrame.
    """
    all_dfs = []

    # Collect files recursively
    files = [
        os.path.join(root, f)
        for root, _, filenames in os.walk(path)
        for f in filenames
        if fnmatch.fnmatch(f, pattern) and f.lower().endswith(('.csv', '.xlsx', '.json'))
    ]

    for f in files:
        ext = os.path.splitext(f)[1].lower()
        if ext == ".csv":
            df = pd.read_csv(f)
        elif ext == ".xlsx":
            df = pd.read_excel(f)
        elif ext == ".json":
            with open(f, "r", encoding="utf-8") as fh:
                raw = json.load(fh)
            if json_key:
                df = pd.DataFrame(raw[json_key])
            else:
                df = pd.DataFrame(raw)
        else:
            continue
        all_dfs.append(df)

    if not all_dfs:
        raise ValueError(f"No valid files found matching pattern '{pattern}' in {path}")

    return pd.concat(all_dfs, ignore_index=True)

path = r"C:\Users\kaustubh.keny\Downloads\gstin_json3"
df = load_and_concat(path,pattern="taxpayerReturnDetails.json")

df.to_csv("2027_GSTIN_DATA.csv")
```
